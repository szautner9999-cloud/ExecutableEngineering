## Solvability of Linear Systems

Some systems of equations don't have a solution, or they do but it's not unique. 

**Example Problem:**
You are organizing a fundraising event and need to buy chairs and tables. 
- Chairs cost \$20 each 
- Tables cost \$50 each
- You have a budget of \$700 
- You need exactly 20 pieces of furniture

How many chairs and tables should you buy?

## A System with One Solution

Let $c$ and $t$ be the number of chairs and tables respectively.

(1) $20 c + 50 t = 700$ (Budget)  
(2) $  c+t = 20$ (Pieces)  

### Solve Graphically

Since these are linear equations, we can plot them as lines on a graph.

```python
A = [[20, 50], [1, 1]]
b = [700, 20]
fig = exe.visual_solve_2d(A, b, labels=['Chairs (c)', 'Tables (t)'], title='Fundraising Event', x_range=(0, 20))
fig.show()
```

The point where the lines intersect satisfies both equations and is therefore a solution. Since lines only cross once, it is a **unique solution**.

## Algebraic Solution (Elimination)

We can solve this algebraically by eliminating one variable.

Multiply the second equation (2) by 20:  
(3) $20c+20t = 400$

Subtract (3) from (1):  
$(20c + 50t) - (20c + 20t) = 700 - 400$  
$30t = 300$  
$t=10$

Substitute $t=10$ back into (2):  
$c = 10$

## Matrix Formulation

Writing this as a matrix equation ($\mathbf{A} \mathbf{x} = \mathbf{b}$):

$$
\begin{pmatrix}
20 & 50 \\
1 &1
\end{pmatrix}  \begin{pmatrix}
c \\
t
\end{pmatrix} =
\begin{pmatrix}
700 \\
20
\end{pmatrix}
$$

Let's find the inverse $\mathbf{A}^{-1}$ such that $\mathbf{x} = \mathbf{A}^{-1}\mathbf{b}$. For a 2x2 matrix:

$$
\begin{pmatrix} a & b \\ c & d \end{pmatrix}^{-1} = \frac{1}{|\mathbf{A}|} \begin{pmatrix} d & -b \\ -c & a \end{pmatrix}
$$

where $|\mathbf{A}| = ad-bc$ is the **determinant**.

For our matrix, $|\mathbf{A}| = 20(1) - 50(1) = -30$. 

$$
\mathbf{A}^{-1} = \frac{1}{-30} \begin{pmatrix} 1 & -50 \\ -1 & 20 \end{pmatrix}
$$

Thus, $\mathbf{x} = \mathbf{A}^{-1} \mathbf{b} = \begin{pmatrix} 10 \\ 10 \end{pmatrix}$.

## Infinite Solutions

Let's tweak our problem. 
- There is a discount on tables: they now cost \$20 each. 
- Your budget was cut to \$400.

The problem is now:  
$20 c + 20 t = 400$  
$  c+t = 20$  

Observe what happens with the lines - what does this mean?

```python
A = [[20, 20], [1, 1]]
b = [400, 20]
fig = exe.visual_solve_2d(A, b, labels=['Chairs (c)', 'Tables (t)'], title='Fundraising Event (Revised)', x_range=(0, 20))
fig.show()
```

### Elimination fails

If we multiply the second equation by 20:  
$20c+20t = 400$

Subtracting the first, we get:  
$0=0$

Any combination of tables and chairs that sums to 20 is a solution. There are an **infinite number of solutions**.

## Singular Matrices

What happens to our matrix formulation?

$$
\mathbf{A} = \begin{pmatrix} 20 & 20 \\ 1 & 1 \end{pmatrix}
$$

The determinant is $|\mathbf{A}| = 20(1) - 20(1) = 0$.

Since the formula for the inverse requires dividing by the determinant ($\frac{1}{|\mathbf{A}|}$), **the inverse does not exist**. 

A matrix with a zero determinant is called **singular**.

## No Solutions

Let's tweak the problem one last time. 
- Tables and chairs both cost \$20.
- The budget is raised to \$500.

$20 c + 20 t = 500$  
$  c+t = 20$

What does the plot imply?

```python
A = [[20, 20], [1, 1]]
b = [500, 20]
fig = exe.visual_solve_2d(A, b, labels=['Chairs (c)', 'Tables (t)'], title='Fundraising Event (Revised Again)', x_range=(0, 20))
fig.show()
```


### The Determinant

Just like the infinite solutions case, the determinant of the coefficient matrix is still zero ($|\mathbf{A}| = 0$). The matrix is singular, and the system cannot be solved exactly.

## Putting it together

Linear equations (in 2 unkowns) are lines in 2D and the solution is the intersection of those lines. 2 Lines can intersect either:

1. in one place which gives you a *unique* solution.
2. everywhere, which gives you *infinite* solutions.
3. nowhere, where there is no solution! 

For 2 and 3, these lines are parallel, i.e. you can slide one to lie ontop of the other.Such lines are called **linearly dependent**. Example 1 has **linear independent** equations which intersect in one place and can be solved.

Scenarios 1 and 2 are called a **consistent** linear system since *an* answer can be obtained. Scenario 3 is **inconsistent** since there is no solution.


## The matrix interpretation

Linear systems generally can be written as,

$$
\bf{A}x = b
$$

where the coefficient matrix $A$ depends on *coefficients* of the unkowns. The constants are strictly in $b$. 

**Solvability**, i.e. the ability to find a unique solution, depends soley on $A$. 

If $|A| \neq 0$, the matrix rows are linearly independent, the matrix is **invertible** (or **non-singular**), and the system has exactly one solution.

If $|A| = 0$, the matrix rows are linearly dependent and the matrix is **non-invertible** (or **singular**). The lines are parallel, which means that there is no *unique* $x$, either you have infinite solutions or none depending on $b$. 

## Approaching singularity

The matrix interpretation is troubling. The solvability of these systems is related to $det(A)$ and we have 2 data points:

1. $|A| = -30$ and the system is solvable.
2. $|A| = 0$ and the system is unsolvable.

What is the significance of $-30$?
What happens when $|A| \rightarrow 0$ ?

We saw $|A| \rightarrow 0$ when we adjusted the price of tables. Let's play with that! Set the cost of tables to $20-\epsilon$.

$20 c + [20+\epsilon] t = 700$

$ c + t = 20$

from which we can find, 

$|A| = 20-20+\epsilon = \epsilon$

which is obviously problematic for small $\epsilon$

If we say $c = 20-t$ and substitute into the first equation:

$ 20 [20 -t] + [20-\epsilon] t = 700$

$\epsilon t= 300$

$ t = 300 / \epsilon$.

Playing with $\epsilon$ we get,

| $\epsilon$ | t | c |
|---|---|---|
| -30 | 10 | 10 |
| -10 | 30 | -10 |
| -1 | 300 | -280 |
| 0.1 | 3000 | -2980 |
| 1e-20 | 3e+22 | -3e+22 |
| 0 |  $\infty$ | $-\infty$ |

which shows the solution is both increasing but also with an increasing rate! 

This should sound alarm bells!

1. Recall $\epsilon \sim 10^{-20}$ is well within the range of roundoff error. Here, even just *recording*  the matrix element value can cause astronomical changes in the solution!
2. These coefficients typically have some uncertainty which compounds the numberical uncertainty.
